## 5. Data quality investigation

Before modelling we check three typical problems: **duplicate rows**, **inconsistent labels** and **outliers**.

### 5.1 Duplicate observations

`df.duplicated()` marks a row as a duplicate when an identical row appeared earlier in the table.

```python
df.duplicated().sum()
```

**Result:** there are **17 duplicated rows**, They should be **removed** with `df.drop_duplicates()`. 

### 5.2 Inconsistent labels in the text columns

Typos or different spellings of the same category (`"yes"`, `"Yes"`, `"y"`) would create fake categories. 

```python
# Distinct values of each text column (NaN excluded): look for typos or unexpected labels
for col in df.select_dtypes(exclude="number").columns:
    print(f"{col:<11}", sorted(df[col].dropna().unique()))
```

### 5.3 Outliers in `absences`


```python
# Boxplot of absences: points beyond the right whisker are potential outliers
plt.figure(figsize=(8, 3))
sns.boxplot(data=df, x="absences")
plt.title("Absences")
plt.show()

# Upper threshold with the IQR rule
q1, q3 = df["absences"].quantile([0.25, 0.75])
upper = q3 + 1.5 * (q3 - q1)
print(f"Upper threshold: {upper:.1f} absences")
print("Students above the threshold:", (df["absences"] > upper).sum())

# The most extreme students, with their grades
df[df["absences"] > upper][["absences", "G1", "G2", "G3"]].sort_values("absences", ascending=False).head(10)
```

**Interpretation:** from the quartiles shown by `describe()` (Q1 = 0, Q3 = 8) the upper threshold is 8 + 1.5 x 8 = **20 absences**, and the maximum of 75 is far beyond it. These values are not impossible (a student can miss many classes), so they should **not be deleted automatically**.

## 7. Initial observations and conclusions

```python
# Compact data-quality summary computed from the data: useful as a final check before preprocessing
summary = pd.Series({
    "rows": len(df),
    "columns": df.shape[1],
    "numeric columns": len(df.select_dtypes(include="number").columns),
    "categorical columns": len(df.select_dtypes(exclude="number").columns),
    "duplicate rows": int(df.duplicated().sum()),
    "total missing values": int(df.isnull().sum().sum()),
    "columns with missing values": int((df.isnull().sum() > 0).sum()),
    "missing values in target G3": int(df["G3"].isnull().sum()),
    "students with G3 = 0": int((df["G3"] == 0).sum()),
})
summary
```

### 7.1 Main conclusions

1. **The dataset is usable but not clean.** It contains duplicates and missing values in every column, but no impossible values or inconsistent labels were found in the statistics we looked at.
2. **The target has two populations.** Most students have grades between 8 and 14 (peaks near 10 and 13) while a distinct group has a grade of 0. A single regression model will fit the zeros badly, so they need a specific decision (keep, flag, model separately or remove).
3. **Grades are linked over time.** Averages decrease slightly from G1 to G3 and the spread increases. `G1` and `G2` are expected to dominate any prediction of `G3`, so we must decide whether to build a model with them (accurate but less useful) or without them (harder but more useful for early intervention).
4. **Most numeric columns are really ordinal scales** (1 to 5 questionnaires, education levels, counts). They should be treated accordingly, and not as continuous measurements.
5. **Several categorical variables are imbalanced** (`higher`, `Pstatus`, `school`, `schoolsup`, `internet`). Results for the small groups are less reliable.
6. **`absences` is right-skewed with extreme values** and needs a decision (keep, cap or log transform).
